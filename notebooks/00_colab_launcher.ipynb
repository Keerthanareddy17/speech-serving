```python
GH_USER = "Keerthanareddy17"
REPO = "speech-serving"

import subprocess, os
from google.colab import drive, userdata
from huggingface_hub import login

# 1) GPU check
print(subprocess.run(["nvidia-smi"], capture_output=True, text=True).stdout)

# 2) Drive for big files only
drive.mount("/content/drive")
os.makedirs("/content/drive/MyDrive/speech-infra", exist_ok=True)

# 3) Hugging Face login: saves the token to ~/.cache/huggingface,
#    so scripts run from the terminal can download Orpheus too
login(token=userdata.get("HF_TOKEN"))

# 4) Clone the repo using the scoped GitHub token
token = userdata.get("GH_TOKEN")
os.chdir("/content")
if not os.path.exists(REPO):
    subprocess.run(["git", "clone", f"https://{token}@github.com/{GH_USER}/{REPO}.git"], check=True)
os.chdir(f"/content/{REPO}")

# 5) Git identity for commits
subprocess.run(["git", "config", "--global", "user.name", "Keerthana Reddy Katasani"])
subprocess.run(["git", "config", "--global", "user.email", "katasanikeerthanareddy17@gmail.com"])

print("ready:", os.getcwd())
print(subprocess.run(["git", "log", "--oneline", "-3"], capture_output=True, text=True).stdout)
```
